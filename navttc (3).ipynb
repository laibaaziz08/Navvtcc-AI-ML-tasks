# Task Thirty Two

```python
import numpy as np

print("--- 1. COPY EXAMPLE ---")
# A copy is a brand new array. Changes to the original do NOT affect the copy.
arr_copy = np.array([1, 2, 3, 4, 5])
x = arr_copy.copy()
arr_copy[0] = 42

print("Original:", arr_copy)
print("Copy (x):", x)
print("-" * 30)

print("--- 2. VIEW EXAMPLE (Modifying Original) ---")
# A view shares the data. Changes to the original DO affect the view.
arr_view = np.array([1, 2, 3, 4, 5])
y = arr_view.view()
arr_view[0] = 42

print("Original:", arr_view)
print("View (y):", y)
print("-" * 30)

print("--- 3. VIEW EXAMPLE (Modifying View) ---")
# Changes made to the view also affect the original array.
arr_view2 = np.array([1, 2, 3, 4, 5])
z = arr_view2.view()
z[0] = 31

print("Original:", arr_view2)
print("View (z):", z)
print("-" * 30)

print("--- 4. CHECKING BASE ATTRIBUTES ---")
# .base returns None if it owns the data (copy), or the original array if it's a view.
sample_arr = np.array([1, 2, 3, 4, 5])
c = sample_arr.copy()
v = sample_arr.view()

print("Copy base (should be None):", c.base)
print("View base (should show original array):", v.base)
```

```python
import pandas as pd

dirty = pd.DataFrame({
    "Name": ["Ali", "Sara", "Ahmed", "Sara", "Zain", "Hina"],
    "Age": [25, 30, None, 30, 220, 28],
    "Date": ["2024/01/05", "2024/02/10", "2024/03/15", "2024/02/10", "wrong", "2024/05/20"],
    "Duration": [45, 60, 50, 60, 45, None],
})
print("Original:\n", dirty)

# 1. Empty cells
print("\nNull counts:\n", dirty.isnull().sum())
clean = dirty.copy()
clean["Age"] = clean["Age"].fillna(clean["Age"].median())
clean["Duration"] = clean["Duration"].fillna(clean["Duration"].mean())

# 2. Wrong format
clean["Date"] = pd.to_datetime(clean["Date"], errors="coerce")
clean = clean.dropna(subset=["Date"])

# 3. Wrong data (age 220 is unrealistic)
clean.loc[clean["Age"] > 100, "Age"] = 100

# 4. Duplicates
print("\nDuplicated rows:", clean.duplicated().sum())
clean = clean.drop_duplicates().reset_index(drop=True)

print("\nCleaned:\n", clean)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(2)
hours = rng.uniform(1, 10, 50)
data = pd.DataFrame({
    "Hours": hours,
    "Score": 40 + 5 * hours + rng.normal(0, 5, 50),
    "Sleep": rng.uniform(4, 9, 50),
    "Noise": rng.normal(0, 1, 50),
})

corr = data.corr()
print(corr.round(2))

plt.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar()
plt.xticks(range(len(corr)), corr.columns, rotation=45)
plt.yticks(range(len(corr)), corr.columns)
plt.title("Correlation matrix")
plt.show()
```

Task Fifty One

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Synthetic data: hours studied -> exam score
rng = np.random.default_rng(42)
hours = rng.uniform(1, 10, 100).reshape(-1, 1)
score = 35 + 6 * hours.ravel() + rng.normal(0, 4, 100)

X_train, X_test, y_train, y_test = train_test_split(
    hours, score, test_size=0.2, random_state=0)

model = LinearRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
print("Intercept:", round(model.intercept_, 2))
print("Slope    :", round(model.coef_[0], 2))
print("MAE :", round(mean_absolute_error(y_test, y_pred), 2))
print("RMSE:", round(np.sqrt(mean_squared_error(y_test, y_pred)), 2))
print("R2  :", round(r2_score(y_test, y_pred), 3))

plt.scatter(X_test, y_test, label="Actual")
plt.plot(X_test, y_pred, color="red", label="Predicted")
plt.xlabel("Hours studied")
plt.ylabel("Score")
plt.legend()
plt.show()
```

Task Fifty Two

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

digits = load_digits()
X_train, X_test, y_train, y_test = train_test_split(
    digits.data, digits.target, test_size=0.25, random_state=0)

model = LogisticRegression(max_iter=5000)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
print("Accuracy:", round(accuracy_score(y_test, y_pred), 4))
print("\nConfusion matrix:\n", confusion_matrix(y_test, y_pred))
print("\n", classification_report(y_test, y_pred))

# Show a few predictions
fig, axes = plt.subplots(1, 5, figsize=(10, 3))
for ax, img, pred in zip(axes, X_test[:5], y_pred[:5]):
    ax.imshow(img.reshape(8, 8), cmap="gray")
    ax.set_title(f"Pred: {pred}")
    ax.axis("off")
plt.show()
```
